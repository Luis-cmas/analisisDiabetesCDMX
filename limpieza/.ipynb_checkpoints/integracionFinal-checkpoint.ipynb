```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
dgis = pd.read_csv("dgis_info.csv")
iter_final = pd.read_csv("./iter_interpolado.csv")
pobreza_final = pd.read_csv("./pobreza_municipal_clean_interpolado.csv")

#dgis.head
dgis[dgis["año"] == 2015]
```

```python
iter_final[iter_final["año"] == 2015]
```

```python
iter_final = iter_final.rename(columns={"mun": "clave_municipio"})
iter_final
```

```python
pobreza_final = pobreza_final[pobreza_final["año"] >= 2015]
pobreza_final
```

```python
union = pd.merge(iter_final, dgis, on=["año", "clave_municipio"], how="outer")
union.columns
```

```python
union
```

```python
union = pd.merge(union, pobreza_final, on=["año", "clave_municipio"], how="outer")
union.columns
```

```python
union
```

```python
# 1. Identificamos las columnas que queremos sumar (todas menos las llaves)
columnas_datos = [col for col in union.columns if col not in ['año', 'clave_municipio', 'nom_mun', 'pobtot', 'pobmas', 'pobfem',
       'p15ym_an_m', 'p15ym_an_f', 'p15ym_se', 'p15ym_se_m', 'p15ym_se_f',
       'p15pri_in', 'p15pri_inm', 'p15pri_inf', 'p15pri_co', 'p15pri_com',
       'p15pri_cof', 'p15sec_in', 'p15sec_inm', 'p15sec_inf', 'p15sec_co',
       'p15sec_com', 'p15sec_cof', 'p18ym_pb', 'p18ym_pb_m', 'p18ym_pb_f',
       'p_0a14', 'p_0a14_m', 'p_0a14_f', 'p_15a24', 'p_15a24_m', 'p_15a24_f',
       'p_25a59', 'p_25a59_m' ]]

# 2. Creamos un DataFrame temporal con las sumas anuales
# Esto calcula el total de todas las alcaldías por año
totales_anuales = union.groupby('año')[columnas_datos].transform('sum')

# 3. Solo aplicamos esa suma a las filas que representan el "Total CDMX"
# Supongamos que la clave del total es '0'
es_fila_total = union['clave_municipio'] == 0 

# Llenamos solo en esas filas los valores calculados
union.loc[es_fila_total, columnas_datos] = totales_anuales[es_fila_total]

```

```python

union = union.drop(columns=['municipio'])
```

```python
union[union["año"] == 2015]
```

```python
# 1. Aseguramos que la máscara sea correcta
es_fila_total = union['clave_municipio'] == 0
columnas_porcentaje = ['pobreza_porcentaje', 'pobreza_extrema_porcentaje', 'pobreza_moderada_porcentaje']
columnas_poblacion = ['pobreza_poblacion', 'pobreza_extrema_poblacion', 'pobreza_moderada_poblacion']

# 2. Realizamos el cálculo usando .values para evitar errores de alineación
numerador = union.loc[es_fila_total, columnas_poblacion]
denominador = union.loc[es_fila_total, 'pobtot']

# Verificamos si el denominador no es cero para evitar errores
if (denominador != 0).all():
    # .values convierte el resultado en una matriz simple de números, 
    # quitando el problema de los índices de Pandas.
    union.loc[es_fila_total, columnas_porcentaje] = (numerador.div(denominador, axis=0) * 100).values
else:
    print("¡Cuidado! Tienes un cero en la población total de la fila seleccionada.")
```
