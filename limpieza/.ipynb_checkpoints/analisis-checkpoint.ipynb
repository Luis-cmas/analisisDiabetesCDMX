```python
import pandas as pd
import matplotlib.pyplot as plt

# Agrupamos para contar casos por municipio y año
casos_por_mun = data.groupby(['nom_mun', 'año']).size().reset_index(name='num_casos')

# Obtenemos la población total (promedio o valor único) por municipio y año
poblacion_por_mun = data.groupby(['nom_mun', 'año'])['pobtot'].mean().reset_index()

# Unimos ambos DataFrames
razon_df = pd.merge(casos_por_mun, poblacion_por_mun, on=['nom_mun', 'año'], how='inner')

# Calculamos la razón o tasa (por cada 100,000 habitantes para facilitar interpretación)
razon_df['tasa_por_100mil'] = (razon_df['num_casos'] / razon_df['pobtot']) * 100000

plt.figure(figsize=(12,6))

for mun in razon_df['nom_mun'].unique():
    subset = razon_df[razon_df['nom_mun'] == mun]
    plt.plot(subset['año'], subset['tasa_por_100mil'], marker='.', label=mun)

plt.title('Tasa de casos por 100,000 habitantes — Todas las alcaldías')
plt.xlabel('Año')
plt.ylabel('Tasa (por 100,000 habitantes)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## clasificacion de dmt

```python
data2= data

data2['clase_enf'] = data2['causa_def'].str.extract(r'(\d+)').astype(int)

def clasificar_diabetes(codigo):
    if 100 <= codigo <= 109:
        return 1
    elif 110 <= codigo <= 119:
        return 2
    elif 120 <= codigo <= 129:
        return 3 #'DM asociada a desnutrición (E12)'
    elif 130 <= codigo <= 139:
        return 4 #'Otras diabetes (E13)'
    elif 140 <= codigo <= 149:
        return 5 #'Diabetes no especificada (E14)'
    else:
        return 'Fuera de rango'

data2['dmt'] = data2['clase_enf'].apply(clasificar_diabetes)

data2[['causa_def','clase_enf','dmt']].head(15)
```

```python
resumen = data['dmt'].value_counts()
resumen.plot(kind='bar', figsize=(10,6))
plt.title('Casos por tipo de diabetes (CIE-10)')
plt.xlabel('Tipo de diabetes')
plt.ylabel('Número de casos')
#plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
ninos = data2[(data2["sexo"] == 1) & (data2["edad"] < 15)]
poblacion_por_anio = (
    ninos.groupby('año')['p_0a14_m']
    .mean()
    .reset_index()
)
casos_por_anio = (
    ninos.groupby('año')
    .size()
    .reset_index(name='num_casos')
)
df_grafica = pd.merge(
    casos_por_anio,
    poblacion_por_anio,
    on='año',
    how='inner'
)
df_grafica['tasa_por_100mil'] = (
    df_grafica['num_casos'] / df_grafica['p_0a14_m']
) * 100000
plt.figure(figsize=(10,5))
plt.plot(df_grafica['año'], df_grafica['tasa_por_100mil'], marker='o', color='darkred')
plt.title('Tasa de casos en menores de 15 años (niños)')
plt.xlabel('Año')
plt.ylabel('Tasa por 100,000 población 0–14')
plt.grid(alpha=0.3)
plt.show()
```

```python
df_hombres_15mas = data[
    (data['sexo'] == 1) & 
    (data['edad'] > 14)
]
casos_por_anio = (
    df_hombres_15mas
    .groupby('año')
    .size()
    .reset_index(name='num_casos')
)
poblacion_por_anio = (
    df_hombres_15mas
    .groupby('año')[['pobmas', 'p_0a14_m']]
    .mean()
    .reset_index()
)
poblacion_por_anio['pob_mayor_14_m'] = (
    poblacion_por_anio['pobmas'] - poblacion_por_anio['p_0a14_m']
)
df_tasa = pd.merge(
    casos_por_anio,
    poblacion_por_anio,
    on='año',
    how='inner'
)
df_tasa['tasa_por_100mil'] = (
    df_tasa['num_casos'] / df_tasa['pob_mayor_14_m']
) * 100000

plt.figure(figsize=(10,5))
plt.plot(
    df_tasa['año'],
    df_tasa['tasa_por_100mil'],
    marker='o'
)

plt.title('Tasa de casos en hombres mayores de 14 años')
plt.xlabel('Año')
plt.ylabel('Tasa por 100,000 hombres (>14)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df_anual = (
    data2
    .groupby('año')
    .agg(
        num_casos=('causa_def', 'count'),
        pobreza_poblacion=('pobreza_moderada_poblacion', 'mean')
    )
    .reset_index()
    .sort_values('año')
)

fig, ax1 = plt.subplots(figsize=(10,5))

# Eje izquierdo: número de casos
ax1.plot(
    df_anual['año'],
    df_anual['num_casos'],
    marker='o',
    label='Número de casos'
)
ax1.set_xlabel('Año')
ax1.set_ylabel('Número de casos')
ax1.grid(alpha=0.3)

# Eje derecho: población en pobreza
ax2 = ax1.twinx()
ax2.plot(
    df_anual['año'],
    df_anual['pobreza_poblacion'],
    marker='s',
    linestyle='--',
    label='Población en pobreza'
)
ax2.set_ylabel('Población en pobreza')

# Título
plt.title('Evolución de casos y población en pobreza')

# Leyenda combinada
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper left')

plt.tight_layout()
plt.show()
```
