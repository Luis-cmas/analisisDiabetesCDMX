```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("transformacion.csv")
data
```

```python
data.columns
```

## alcaldia-pobreza

```python
df_modelo_pobreza = data.groupby(['municipio','año']).agg({
    'pobtot': 'first',
    'pobreza_porcentaje': 'first',
    'pobreza_extrema_porcentaje': 'first',
    'pobreza_moderada_porcentaje': 'first',
    'p_0a14': 'first',
    'p_15a24': 'first',
    'p_25a59': 'first',
    'p_60ymas': 'first',
}).reset_index()
df_modelo_pobreza
```

```python
df_modelo_pobreza['pct_0a14'] = df_modelo_pobreza['p_0a14'] / df_modelo_pobreza['pobtot']
df_modelo_pobreza['pct_15a24'] = df_modelo_pobreza['p_15a24'] / df_modelo_pobreza['pobtot']
df_modelo_pobreza['pct_60ymas'] = df_modelo_pobreza['p_60ymas'] / df_modelo_pobreza['pobtot']
```

```python
features = [
    'pobreza_porcentaje',
    'pobreza_extrema_porcentaje',
    'pct_0a14',
    'pct_15a24',
    'pct_60ymas'
]
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt

X = df_modelo_pobreza[features]
X_scaled = StandardScaler().fit_transform(X)

inercia = []
K = range(2, 16)

for k in K:
    kmeans_pobreza = KMeans(n_clusters=k, random_state=42)
    kmeans_pobreza.fit(X_scaled)
    inercia.append(kmeans_pobreza.inertia_)

plt.figure(figsize=(6,4))
plt.plot(K, inercia, marker='o')
plt.xlabel('Número de clusters (k)')
plt.ylabel('Inercia')
plt.title('Método del Codo')
plt.grid(alpha=0.3)
plt.show()

```

```python
from sklearn.metrics import silhouette_score

silhouette = []

for k in range(2, 16):
    kmeans_pobreza = KMeans(n_clusters=k, random_state=42)
    labels = kmeans_pobreza.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette.append(score)

plt.figure(figsize=(6,4))
plt.plot(range(2,16), silhouette, marker='o')
plt.xlabel('Número de clusters (k)')
plt.ylabel('Silhouette score')
plt.title('Silhouette score por k')
plt.grid(alpha=0.3)
plt.show()
```

```python
kmeans_pobreza = KMeans(n_clusters=5, random_state=42)
#    labels = kmeans.fit_predict(X_scaled)
df_modelo_pobreza['cluster'] = kmeans_pobreza.fit_predict(X_scaled)
```

```python
print(df_modelo_pobreza.groupby('municipio')['cluster'].nunique())
```

```python

conteo = df_modelo_pobreza.groupby('cluster')['municipio'].nunique()

conteo.plot(kind='bar', figsize=(6,4))
plt.title('Número de alcaldías por cluster')
plt.xlabel('Cluster')
plt.ylabel('Número de alcaldías')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))

for c in sorted(df_modelo_pobreza['cluster'].unique()):
    subset = df_modelo_pobreza[df_modelo_pobreza['cluster'] == c]
    plt.scatter(
        subset['pobreza_porcentaje'],
        subset['pct_0a14'],
        label=f'Cluster {c}',
        alpha=0.7
    )

plt.xlabel('Pobreza (%)')
plt.ylabel('% Población 0–14')
plt.title('Alcaldías agrupadas por cluster')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

cluster_profile = df_modelo_pobreza.groupby('cluster')[features].mean()

plt.figure(figsize=(8,5))
sns.heatmap(cluster_profile, annot=True, cmap='viridis')
plt.title('Perfil promedio de cada cluster')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

for mun in df_modelo_pobreza['municipio'].unique():
    subset = df_modelo_pobreza[df_modelo_pobreza['municipio'] == mun]
    plt.plot(subset['año'], subset['cluster'], alpha=0.4)

plt.xlabel('Año')
plt.ylabel('Cluster')
plt.title('Evolución de alcaldías entre clusters')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## modelo para alcaldia-escolaridad

```python
df_esc = (
    data
    .groupby(['municipio', 'escolarida'])
    .size()
    .unstack(fill_value=0)
)
```

```python
df_esc_pct = df_esc.div(df_esc.sum(axis=1), axis=0)
```

```python
X = StandardScaler().fit_transform(df_esc_pct)
```

```python
kmeans_escolaridad = KMeans(n_clusters=4, random_state=42)
clusters = kmeans_escolaridad.fit_predict(X)

df_esc_pct['cluster'] = clusters
```

```python
perfil_clusters = df_esc_pct.groupby('cluster').mean()
print(perfil_clusters)
```

```python
df_esc_pct.reset_index().groupby('cluster')['municipio'].apply(list)
```

```python
perfil_clusters.T.plot(
    kind='bar',
    figsize=(12,6)
)

plt.title('Perfil de escolaridad por cluster')
plt.ylabel('Proporción de casos')
plt.xlabel('Nivel educativo')
plt.legend(title='Cluster')
plt.tight_layout()
plt.show()
```

```python
inercia = []

K = range(2, 11)

for k in K:
    kmeans_escolaridad = KMeans(n_clusters=k, random_state=42)
    kmeans_escolaridad.fit(X)
    inercia.append(kmeans_escolaridad.inertia_)

plt.figure(figsize=(8,5))
plt.plot(K, inercia, marker='o')
plt.xlabel('Número de clusters')
plt.ylabel('Inercia')
plt.title('Método del Codo')
plt.show()
```

```python
sil_scores = {}

for k in range(2, 11):
    kmeans_escolaridad = KMeans(n_clusters=k, random_state=42)
    labels = kmeans_escolaridad.fit_predict(X)
    sil_scores[k] = silhouette_score(X, labels)

sil_scores
```
