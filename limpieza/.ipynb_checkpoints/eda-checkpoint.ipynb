```python

# 1. Crear el DataFrame destino
df_tasas_panel = pd.DataFrame()

# 2. Conservar identificadores de tiempo y espacio
df_tasas_panel['año'] = union['año']
df_tasas_panel['nom_mun'] = union['nom_mun']

# --- TASAS DE MORTALIDAD GENERAL Y POR SEXO ---
# Tasa general (por 10,000 habs.)
df_tasas_panel['tasa_mortalidad'] = (
    (union['dmt1_total'] + union['dmt2_total']) / union['pobtot']
) * 100000

# Tasa masculina (por 10,000 hombres)
df_tasas_panel['tasa_mortalidad_masc'] = (
    (union['dmt1_masc'] + union['dmt2_masc']) / union['pobmas']
) * 100000

# Tasa femenina (por 10,000 mujeres)
df_tasas_panel['tasa_mortalidad_fem'] = (
    (union['dmt1_fem'] + union['dmt2_fem']) / union['pobfem']
) * 100000

# --- POBREZA ---
df_tasas_panel['pobreza_porcentaje'] = union['pobreza_porcentaje']

# --- TASAS POR GRUPOS DE EDAD (por 10,000 habs. del grupo) ---
# Grupo 0 a 14 años
df_tasas_panel['tasa_edad_0_14'] = (
    (
        union['0-14_sexo1_dmt1']
        + union['0-14_sexo1_dmt2']
        + union['0-14_sexo2_dmt1']
        + union['0-14_sexo2_dmt2']
    )
    / union['p_0a14']
) * 100000

# Grupo 15 a 24 años
df_tasas_panel['tasa_edad_15_24'] = (
    (
        union['15-24_sexo1_dmt1']
        + union['15-24_sexo1_dmt2']
        + union['15-24_sexo2_dmt1']
        + union['15-24_sexo2_dmt2']
    )
    / union['p_15a24']
) * 100000

# Grupo 25 a 59 años
df_tasas_panel['tasa_edad_25_59'] = (
    (
        union['25-59_sexo1_dmt1']
        + union['25-59_sexo1_dmt2']
        + union['25-59_sexo2_dmt1']
        + union['25-59_sexo2_dmt2']
    )
    / union['p_25a59']
) * 100000

# Grupo 60 años y más
df_tasas_panel['tasa_edad_60mas'] = (
    (
        union['60+_sexo1_dmt1']
        + union['60+_sexo1_dmt2']
        + union['60+_sexo2_dmt1']
        + union['60+_sexo2_dmt2']
    )
    / union['p_60ymas']
) * 100000

# --- TASAS POR ESCOLARIDAD (por 10,000 habs. de cada nivel) ---
# Sin escolaridad / Preescolar
casos_sin_esc_pre = (
    union['hombres_dmt1_sin_esc']
    + union['hombres_dmt1_preescolar']
    + union['hombres_dmt2_sin_esc']
    + union['hombres_dmt2_preescolar']
    + union['mujeres_dmt1_sin_esc']
    + union['mujeres_dmt1_preescolar']
    + union['mujeres_dmt2_sin_esc']
    + union['mujeres_dmt2_preescolar']
)
df_tasas_panel['tasa_esc_sin_preescolar'] = (
    casos_sin_esc_pre / union['p15ym_se']
) * 100000

# Primaria
casos_primaria = (
    union['hombres_dmt1_primaria']
    + union['hombres_dmt2_primaria']
    + union['mujeres_dmt1_primaria']
    + union['mujeres_dmt2_primaria']
)
pob_primaria = union['p15pri_in'] + union['p15pri_co']
df_tasas_panel['tasa_esc_primaria'] = (casos_primaria / pob_primaria) * 100000

# Secundaria
casos_secundaria = (
    union['hombres_dmt1_secundaria']
    + union['hombres_dmt2_secundaria']
    + union['mujeres_dmt1_secundaria']
    + union['mujeres_dmt2_secundaria']
)
pob_secundaria = union['p15sec_in'] + union['p15sec_co']
df_tasas_panel['tasa_esc_secundaria'] = (
    casos_secundaria / pob_secundaria
) * 100000

# Preparatoria, Profesional y Posgrado
casos_prep_prof_pos = (
    union['hombres_dmt1_preparatoria']
    + union['hombres_dmt1_profesional']
    + union['hombres_dmt1_posgrado']
    + union['hombres_dmt2_preparatoria']
    + union['hombres_dmt2_profesional']
    + union['hombres_dmt2_posgrado']
    + union['mujeres_dmt1_preparatoria']
    + union['mujeres_dmt1_profesional']
    + union['mujeres_dmt1_posgrado']
    + union['mujeres_dmt2_preparatoria']
    + union['mujeres_dmt2_profesional']
    + union['mujeres_dmt2_posgrado']
)
df_tasas_panel['tasa_esc_prep_superior'] = (
    casos_prep_prof_pos / union['p18ym_pb']
) * 100000

# Clean potential division by zero or NaN values
df_tasas_panel = df_tasas_panel.replace([np.inf, -np.inf], np.nan).fillna(0)

# Inspección del DataFrame resultante
print(df_tasas_panel.head())

df_tasas_panel.to_csv("df_tasass.csv", index=False, encoding="utf-8")
```

```python

# 1. Filtramos el último año y quitamos el "Total CDMX" para no sesgar la gráfica
# Suponiendo que la clave 0 es el total
ultimo_anio = union[(union['año'] == union['año'].max()) & (union['clave_municipio'] != 0)].copy()

# 2. Sumamos los casos totales por sexo
ultimo_anio['total_dmt_h'] = ultimo_anio.filter(like='hombres_dmt').sum(axis=1)
ultimo_anio['total_dmt_m'] = ultimo_anio.filter(like='mujeres_dmt').sum(axis=1)
ultimo_anio['total_dmt_general'] = ultimo_anio['total_dmt_h'] + ultimo_anio['total_dmt_m']

# 3. Calculamos la TASA (Casos por cada 100,000 habitantes)
# Usamos pobtot para la normalización
ultimo_anio['tasa_dmt_h'] = (ultimo_anio['total_dmt_h'] / ultimo_anio['pobtot']) * 100000
ultimo_anio['tasa_dmt_m'] = (ultimo_anio['total_dmt_m'] / ultimo_anio['pobtot']) * 100000
ultimo_anio['tasa_total'] = (ultimo_anio['total_dmt_general'] / ultimo_anio['pobtot']) * 100000

# 4. Ordenamos de mayor a menor tasa total
ultimo_anio = ultimo_anio.sort_values('tasa_total', ascending=False)

# 5. Graficamos
plt.figure(figsize=(14, 7))

# Usamos una gráfica de barras apiladas o agrupadas
# Aquí la opción de pandas es sencilla, pero la configuramos para que se vea bien
ax = ultimo_anio.plot(
    x='nom_mun', 
    y=['tasa_dmt_h', 'tasa_dmt_m'], 
    kind='bar', 
    stacked=False,#True, # Apiladas para ver la tasa total arriba
    figsize=(14, 6),
    color=['#3498db', '#e74c3c'] # Azul para hombres, Rojo para mujeres
)

plt.title(f'Tasa de casos DMT por cada 100,000 habitantes por Alcaldía ({ultimo_anio["año"].max()})', fontsize=14)
plt.ylabel('Tasa (Casos / 100k hab)', fontsize=12)
plt.xlabel('Alcaldía', fontsize=12)
plt.legend(['Hombres', 'Mujeres'])
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.savefig('graficas/dmt2020.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python


# Creamos una copia para no alterar el original por error
df_tasas = union.copy()

# Nueva versión de la función más flexible
def calcular_tasa(datos_casos, nombre_col_pob):
    # Si datos_casos es un texto, lo buscamos en el df. 
    # Si ya es una suma/serie, lo usamos directo.
    if isinstance(datos_casos, str):
        casos = df_tasas[datos_casos]
    else:
        casos = datos_casos
        
    return (casos / df_tasas[nombre_col_pob]) * 100000


# 1. Tasa General de DMT (Casos Totales)
df_tasas['tasa_dmt_total'] = calcular_tasa('casos totales', 'pobtot')

# 2. Tasas por Tipo de Diabetes
df_tasas['tasa_dmt1'] = calcular_tasa('dmt1_total', 'pobtot')
df_tasas['tasa_dmt2'] = calcular_tasa('dmt2_total', 'pobtot')

# 3. Tasas por Sexo
df_tasas['tasa_dmt_fem'] = calcular_tasa(df_tasas['dmt1_fem'] + df_tasas['dmt2_fem'], 'pobfem')
df_tasas['tasa_dmt_masc'] = calcular_tasa(df_tasas['dmt1_masc'] + df_tasas['dmt2_masc'], 'pobmas')

# 4. Tasas de Educación (Analfabetismo y rezago)
# Nota: p15ym_an es población de 15 años y más analfabeta
df_tasas['tasa_analfabetismo'] = (df_tasas['p15ym_an'] / df_tasas['pobtot']) * 100
df_tasas
```

```python
# Columnas de DMT en personas con estudios superiores (Licenciatura/Posgrado)
cols_superior = [col for col in df_tasas.columns if 'profesional' in col or 'posgrado' in col]
df_tasas['casos_dmt_estudios_superiores'] = df_tasas[cols_superior].sum(axis=1)
df_tasas['tasa_dmt_estudios_superiores'] = calcular_tasa('casos_dmt_estudios_superiores', 'pobtot')

# Columnas de DMT en personas con poca o nula escolaridad
cols_basica = [col for col in df_tasas.columns if 'sin_esc' in col or 'preescolar' in col or 'primaria' in col]
df_tasas['casos_dmt_estudios_basicos'] = df_tasas[cols_basica].sum(axis=1)
df_tasas['tasa_dmt_estudios_basicos'] = calcular_tasa('casos_dmt_estudios_basicos', 'pobtot')
```

```python

# Seleccionamos las variables normalizadas
variables_analisis = [
    'pobreza_porcentaje', 
    'pobreza_extrema_porcentaje',
    'tasa_dmt_total',
    'tasa_dmt1',
    'tasa_dmt2',
    'tasa_dmt_estudios_basicos',
    'tasa_dmt_estudios_superiores',
    'tasa_analfabetismo'
]

plt.figure(figsize=(10, 8))
sns.heatmap(df_tasas[variables_analisis].corr(), annot=True, cmap='RdYlGn', center=0)
plt.title('Correlación: Pobreza, Salud (DMT) y Educación')
plt.savefig('graficas/correlaciones.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python

# 1. Seleccionamos solo las columnas necesarias y quitamos el total CDMX
df_genero = df_tasas[df_tasas['clave_municipio'] != 0][['año', 'nom_mun', 'tasa_dmt_masc', 'tasa_dmt_fem']].copy()

# 2. Reestructuramos el dataframe (Melt)
df_melted = df_genero.melt(id_vars=['año', 'nom_mun'], 
                           value_vars=['tasa_dmt_masc', 'tasa_dmt_fem'],
                           var_name='Sexo', value_name='Tasa_DMT')

# Limpiamos los nombres para la gráfica
df_melted['Sexo'] = df_melted['Sexo'].replace({'tasa_dmt_masc': 'Hombres', 'tasa_dmt_fem': 'Mujeres'})

# 3. Graficamos
plt.figure(figsize=(14, 7))
sns.boxplot(data=df_melted, x='año', y='Tasa_DMT', hue='Sexo', palette='muted')

plt.title('Comparativa de la Tasa de DMT: Hombres vs Mujeres por Año', fontsize=16)
plt.ylabel('Tasa de Casos (por cada 100k hab)', fontsize=12)
plt.xlabel('Año', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.savefig('graficas/comparativaHyM.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
def obtener_outliers_por_sexo(df, anio, columna_tasa):
    data = df[(df['año'] == anio) & (df['clave_municipio'] != 0)]
    q1 = data[columna_tasa].quantile(0.25)
    q3 = data[columna_tasa].quantile(0.75)
    iqr = q3 - q1
    limite = q3 + 1.5 * iqr
    return data[data[columna_tasa] > limite][['nom_mun', columna_tasa]]

ultimo = 2015
out_h = obtener_outliers_por_sexo(df_tasas, ultimo, 'tasa_dmt_masc')
out_m = obtener_outliers_por_sexo(df_tasas, ultimo, 'tasa_dmt_fem')

print(f"--- OUTLIERS EN HOMBRES ({ultimo}) ---\n", out_h)
print(f"\n--- OUTLIERS EN MUJERES ({ultimo}) ---\n", out_m)
```

```python


# 1. Filtramos el último año y alcaldías (sin total)
df_piramide = df_tasas[(df_tasas['año'] == 2015) & (df_tasas['clave_municipio'] != 0)].copy()

# 2. Definimos los grupos de edad y tipos de DMT que queremos mapear
edades = ['0-14', '15-24', '25-59', '60+']
datos_lista = []

for edad in edades:
    for dmt in [1, 2]:
        # Extraemos valores para hombres (sexo1) y mujeres (sexo2)
        h = df_piramide[f'{edad}_sexo1_dmt{dmt}'].sum()
        m = df_piramide[f'{edad}_sexo2_dmt{dmt}'].sum()
        
        datos_lista.append({'Edad': edad, 'Sexo': 'Hombres', 'DMT': f'Tipo {dmt}', 'Casos': -h}) # Hombres a la izquierda (negativo)
        datos_lista.append({'Edad': edad, 'Sexo': 'Mujeres', 'DMT': f'Tipo {dmt}', 'Casos': m})  # Mujeres a la derecha (positivo)

df_plot = pd.DataFrame(datos_lista)

# 3. Graficamos
g = sns.FacetGrid(df_plot, col="DMT", hue="Sexo", height=6, aspect=0.8, palette="muted", sharex=False)

g.map(sns.barplot, "Casos", "Edad", order=edades)

# Estética de pirámide
for ax in g.axes.flat:
    ax.axvline(0, color='black', lw=1) # Línea central
    # Ajustamos las etiquetas del eje X para que no se vean negativas en el lado de hombres
    ticks = ax.get_xticks()
    ax.set_xticklabels([int(abs(tick)) for tick in ticks])

g.add_legend()
g.set_axis_labels("Número de Casos", "Rango de Edad")
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle('Pirámide de Incidencia: Edad, Sexo y Tipo de Diabetes (CDMX-2015)', fontsize=16)
plt.savefig('graficas/incidenciaEdadSexo.png', dpi=300, bbox_inches='tight')

plt.show()
```

## propuesta de PCA

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import numpy as np

# 1. Seleccionamos las variables que definen el "perfil" de la alcaldía
# Usamos las tasas y porcentajes que ya calculamos
cols_pca = [
    'pobreza_porcentaje', 'pobreza_extrema_porcentaje', 
    'tasa_analfabetismo', 'tasa_dmt_total',
    'p_60ymas', 'p_0a14', 'tasa_dmt_estudios_superiores'
]

# Filtramos el último año y quitamos el total estatal
df_pca_input = df_tasas[(df_tasas['año'] == df_tasas['año'].max()) & (df_tasas['clave_municipio'] != 0)]
nombres_municipios = df_pca_input['nom_mun'].values

# 2. Escalamos los datos
scaler = StandardScaler()
data_scaled = scaler.fit_transform(df_pca_input[cols_pca])


# 3. Aplicamos PCA para 2 componentes (n_components es el nombre correcto)
pca = PCA(n_components=2) 
componentes = pca.fit_transform(data_scaled)

# Creamos un DF con los resultados
df_pca_final = pd.DataFrame(data = componentes, columns = ['Componente 1', 'Componente 2'])
df_pca_final['Alcaldía'] = nombres_municipios
```

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df_pca_final, x='Componente 1', y='Componente 2', s=100)

# Etiquetamos cada punto
for i, txt in enumerate(df_pca_final['Alcaldía']):
    plt.annotate(txt, (df_pca_final['Componente 1'].iloc[i], df_pca_final['Componente 2'].iloc[i]), 
                 xytext=(5,5), textcoords='offset points', fontsize=9)

plt.axhline(0, color='gray', linestyle='--', alpha=0.5)
plt.axvline(0, color='gray', linestyle='--', alpha=0.5)
plt.title('Mapa de Perfiles de las Alcaldías (PCA)', fontsize=15)
plt.xlabel(f'Componente 1 ({pca.explained_variance_ratio_[0]*100:.1f}% de varianza)')
plt.ylabel(f'Componente 2 ({pca.explained_variance_ratio_[1]*100:.1f}% de varianza)')
plt.savefig('graficas/mapaPerfiles.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Ver la influencia de cada variable
loadings = pd.DataFrame(pca.components_.T, columns=['PC1', 'PC2'], index=cols_pca)
print("Influencia de las variables en los componentes:")
print(loadings.sort_values(by='PC1', ascending=False))
```
