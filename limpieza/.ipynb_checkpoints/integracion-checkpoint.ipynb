```python
import pandas as pd
from datetime import datetime
import matplotlib.pyplot as plt
```

```python
pobreza = pd.read_csv("pobreza_municipal_clean.csv",parse_dates = ["periodo"])
pobreza
```

```python
pobreza_ord = pobreza[["municipio","pobreza_poblacion","periodo"]].sort_values(by="municipio")
pobreza_ord
pobreza2=pobreza
```

```python
pobreza2_ord = pobreza2.sort_values(by=["municipio","periodo"])
pobreza2_ord
```

```python
pobreza2_ord["entidad_federativa"] = pobreza2_ord["entidad_federativa"].replace("Distrito Federal", "Ciudad de México")
pobreza2_ord
```

```python
pobreza2_ord["año"] = pobreza2_ord["periodo"].dt.year
pobreza2_ord = pobreza2_ord[['clave_municipio', 'municipio','pobreza_porcentaje',
                             'pobreza_poblacion', 'pobreza_extrema_porcentaje','pobreza_extrema_poblacion',
                             'pobreza_moderada_porcentaje','pobreza_moderada_poblacion', 'año']]
pobreza2_ord
```

```python
alcaldias = pobreza2_ord[pobreza2_ord["año"] == 2010]
alcaldias_cdmx = alcaldias["municipio"]
alcaldias_cdmx
```

```python
#anios = [2011,2012,2013,2014,2016,2017,2018,2019]
#for alcaldia in alcaldias_cdmx:
#    for anio in anios:
#        alcaldias_cdmx.iloc[]
#    print(alcaldia)
# Crear una lista vacía para guardar los resultados
dfs = []

# Recorremos cada municipio
for municipio, grupo in pobreza2_ord.groupby('municipio'):
    # Rango completo de años entre el mínimo y máximo
    años_completos = range(grupo['año'].min(), grupo['año'].max() + 1)
    
    # Reindexar el grupo con los años completos
    grupo_completo = grupo.set_index('año').reindex(años_completos).reset_index()
    
    # Rellenar el nombre del municipio en las nuevas filas
    grupo_completo['municipio'] = municipio
    
    # Opcional: si quieres conservar los valores conocidos y dejar NaN en los nuevos
    # puedes usar forward fill (rellenar con último valor conocido)
    # grupo_completo = grupo_completo.ffill()
    
    # Agregar el grupo expandido a la lista
    dfs.append(grupo_completo)

# Unir todos los municipios
df_final = pd.concat(dfs, ignore_index=True)
df_final.rename(columns={'index': 'año'}, inplace=True)

df_final
```

```python
#realizamos la interpolacion para obtener los valores de los años faltantes

df_final = df_final.reset_index(drop=True)

df_final = df_final.sort_values(['municipio', 'año'])
poblaciones=['pobreza_poblacion','pobreza_extrema_poblacion','pobreza_moderada_poblacion','clave_municipio']
porcentajes=['pobreza_porcentaje','pobreza_extrema_porcentaje','pobreza_moderada_porcentaje']
df_final = (
    df_final
    .groupby('municipio', group_keys=False)
    .apply(lambda g: g.assign(**g[poblaciones].interpolate(method='linear').round(0)))
    .reset_index(drop=True)
)
df_final = (
    df_final
    .groupby('municipio', group_keys=False)
    .apply(lambda g: g.assign(**g[porcentajes].interpolate(method='linear')))
    .reset_index(drop=True)
)
# df_final = (
#     df_final
#     .groupby('municipio', group_keys=False)
#     .apply(lambda g: g.interpolate(method='linear').round(0))
#     .reset_index(drop=True)
# )
# df_final[porcentajes] = (
#     df_final
#     .groupby('municipio', group_keys=False)
#     .apply(lambda g: g.interpolate(method='linear'))
#     .reset_index(drop=True)
# )

df_final.tail(10)

```

```python
df_final[df_final["clave_municipio"] == 9002]
```

```python
df_final["clave_municipio"] = df_final["clave_municipio"] - 9000
df_final["clave_municipio"]
```

```python
df_final.to_csv("./pobreza_municipal_clean_interpolado.csv", index=False, encoding="utf-8")
```

```python
# Crear una lista vacía para guardar los resultados
dfs2 = []

# Recorremos cada municipio
for municipio2, grupo2 in iter_ord.groupby('nom_mun'):
    # Rango completo de años entre el mínimo y máximo
    años_completos2 = range(grupo2['año'].min(), grupo2['año'].max() + 1)
    
    # Reindexar el grupo con los años completos
    grupo_completo2 = grupo2.set_index('año').reindex(años_completos2).reset_index()
    
    # Rellenar el nombre del municipio en las nuevas filas
    grupo_completo2['nom_mun'] = municipio2
    
    # Opcional: si quieres conservar los valores conocidos y dejar NaN en los nuevos
    # puedes usar forward fill (rellenar con último valor conocido)
    # grupo_completo = grupo_completo.ffill()
    
    # Agregar el grupo expandido a la lista
    dfs2.append(grupo_completo2)

# Unir todos los municipios
df_final2= pd.concat(dfs2, ignore_index=True)
df_final2.rename(columns={'index': 'año'}, inplace=True)

df_final2
```

```python
#realizamos la interpolacion para obtener los valores de los años faltantes

df_final2 = df_final2.reset_index(drop=True)

df_final2 = df_final2.sort_values(['nom_mun', 'año'])

df_final2 = (
    df_final2
    .groupby('nom_mun', group_keys=False)
    .apply(lambda g: g.interpolate(method='linear').round(0))
    .reset_index(drop=True)
)

df_final2.head(10)
```
