## dataset de DGIS

```python
defunciones2015 = pd.read_csv("defunciones_generales_2015.csv")

defunciones2015.info()
```

```python
defunciones2015.describe
```

```python
defunciones2015
```

```python
defunciones2015_cdmx = defunciones2015[defunciones2015["ent_resid"] == 9]
defunciones2015_cdmx
```

```python
print(list(defunciones2015_cdmx.columns))
```

```python
defunciones2015_cdmx_variables = defunciones2015_cdmx[['ent_resid', 'mun_resid','causa_def', 'sexo', 'edad','edad_agru', 'dia_ocurr', 'mes_ocurr', 'anio_ocur','escolarida' ]]
defunciones2015_cdmx_variables
```

```python
defunciones2015_cdmx_diabetes = defunciones2015_cdmx_variables[(defunciones2015_cdmx_variables["causa_def"] >= "E100") & (defunciones2015_cdmx_variables["causa_def"] <= "E149")] #desde E100-E149 son los casos de diabetes mellitus
defunciones2015_cdmx_diabetes
```

```python
defunciones2015_cdmx_diabetes
```

```python
defunciones2015_cdmx_diabetes.plot.scatter(x="mun_resid",y="edad_agru")
```

```python
#eliminamos los casos no especificados
defunciones2015_cdmx_diabetes_clean = defunciones2015_cdmx_diabetes
defunciones2015_cdmx_diabetes_clean = defunciones2015_cdmx_diabetes_clean[defunciones2015_cdmx_diabetes["mun_resid"] != 999]
defunciones2015_cdmx_diabetes_clean = defunciones2015_cdmx_diabetes_clean[defunciones2015_cdmx_diabetes["edad_agru"] != 30]
defunciones2015_cdmx_diabetes_clean = defunciones2015_cdmx_diabetes_clean[defunciones2015_cdmx_diabetes["escolarida"] != 99]
```

```python
defunciones2015_cdmx_diabetes_clean.plot.scatter(x="mun_resid",y="causa_def")
```

```python
defunciones2015_cdmx_diabetes.info()
```

```python
print(defunciones2015_cdmx_diabetes.isnull().sum())
```

```python
defunciones2015_cdmx_diabetes.to_csv("./limpieza/defunciones2015_clean.csv", index=False, encoding="utf-8")
```

 ## limpieza defunciones 2016

```python
defunciones2016 = pd.read_csv("defunciones_generales_2016.csv")

defunciones2016.info()
```

```python
print(list(defunciones2016.columns))
```

```python
#filtrado de ciudad
defunciones2016_cdmx = defunciones2016[defunciones2016["ent_resid"] == 9]
#filtrado de variables
defunciones2016_cdmx = defunciones2016_cdmx[['ent_resid', 'mun_resid','causa_def', 'sexo', 'edad','edad_agru', 'dia_ocurr', 'mes_ocurr', 'anio_ocur','escolarida' ]]
#filtrado de enfermedad
defunciones2016_cdmx = defunciones2016_cdmx[(defunciones2016_cdmx["causa_def"] >= "E100") & (defunciones2016_cdmx["causa_def"] <= "E149")] #desde E100-E149 son los casos de diabetes mellitus
#eliminamos los casos no especificados
defunciones2016_cdmx = defunciones2016_cdmx[defunciones2016_cdmx["mun_resid"] != 999]
defunciones2016_cdmx = defunciones2016_cdmx[defunciones2016_cdmx["edad_agru"] != 30]
defunciones2016_cdmx = defunciones2016_cdmx[defunciones2016_cdmx["escolarida"] != 99]
defunciones2016_cdmx
```

```python
defunciones2016_cdmx.plot.scatter(x="mun_resid",y="edad_agru")
```

```python
defunciones2016_cdmx.to_csv("./limpieza/defunciones2016_clean.csv", index=False, encoding="utf-8")
```
